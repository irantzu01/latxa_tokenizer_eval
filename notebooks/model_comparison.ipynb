```python
from transformers import AutoTokenizer, AutoModelForCausalLM
from datasets import load_dataset
```

```python
# Load models
latxa7B_tokenizer = AutoTokenizer.from_pretrained("HiTZ/latxa-7b-v1.2")
#latxa7B_model = AutoModelForCausalLM.from_pretrained("HiTZ/latxa-7b-v1.2")
latxa70B_tokenizer = AutoTokenizer.from_pretrained("HiTZ/latxa-70b-v1.2")
#latxa70B_model = AutoModelForCausalLM.from_pretrained("HiTZ/latxa-70b-v1.2")
llama2_tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-2-7b-hf")
#llama2_model = AutoModelForCausalLM.from_pretrained("meta-llama/Llama-2-7b-hf")
```

```python
def analyze_family(words):
    print(f"{'Word':<15} | {'Latxa 7B':<40} | {'Latxa 70B'}")
    print("-" * 90)
    for w in words:
        tokens7b = latxa7B_tokenizer.tokenize(w)
        tokens70b = latxa70B_tokenizer.tokenize(w)

        print(
            f"{w:<15} | "
            f"{str(tokens7b):<40} | "
            f"{tokens70b}"
        )

for family, words in families.items():
    print(f"\n############################")
    print(f"### {family}")
    print("############################")
    analyze_family(words)
```

Latxa 7B and 70B tokenize frequent Basque suffixes exactly the same way.

# 2. Vocabulary-level comparison

### 1.1 Compare tokenizer properties

Tokenizers employ BPE algorithm, using the implementation from SentencePiece.

Llama2 was mostly pre-trained on English data (89.70%). The next biggest (8.38%) group has the unknown label, which is partially made up of programming code data. Then German (0.17%), French (0.16%),... In total, it was pretrained with 2 trillion tokens.

Latxa is based on Llama 2, which they continued pretraining on a new Basque corpus comprising 4.3M documents and 4.2B tokens.

### 1.2 Compare Vocabularies

```python
# Percentage of shared tokens
def shared_tokens_percentage(tokenizer1, tokenizer2):
    vocab1 = set(tokenizer1.get_vocab().keys())
    vocab2 = set(tokenizer2.get_vocab().keys())
    shared = vocab1 & vocab2
    return 100 * len(shared) / len(vocab1)

def unique_tokens(tokenizer1, tokenizer2):
    vocab1 = set(tokenizer1.get_vocab().keys())
    vocab2 = set(tokenizer2.get_vocab().keys())
    only_in_1 = vocab1 - vocab2
    only_in_2 = vocab2 - vocab1
    return only_in_1, only_in_2

unique_latxa7B, unique_latxa70B = unique_tokens(latxa7B_tokenizer, llama2_tokenizer)
print("
print("Shared vocab percentage Latxa 7B and Latxa 70B:", shared_tokens_percentage(latxa7B_tokenizer, latxa70B_tokenizer))
print("Shared vocab percentage Latxa 7B and Llama2:", shared_tokens_percentage(latxa7B_tokenizer, llama2_tokenizer))
```

### Evaluation of EusProficiency

```python
#Load dataset and prepare prompts
ds = load_dataset("HiTZ/EusProficiency", split="test")

def format_prompt(question, candidates):
    return (
        f"Galdera: {question}\n"
        f"A: {candidates[0]}\n"
        f"B: {candidates[1]}\n"
        f"C: {candidates[2]}\n"
        f"D: {candidates[3]}\n"
        f"Erantzuna:"
    )

CHOICES = [" A", " B", " C", " D"]
evaluation_items = []
for item in ds:
    prompt = format_prompt(item["question"], item["candidates"])
    choice_texts = [prompt + choice for choice in CHOICES]

    evaluation_items.append({
        "choice_texts": choice_texts,  # list[str] length 4
        "answer": item["answer"]       # int 0–3
    })
print("Evaluation items prepared:", len(evaluation_items))
```
